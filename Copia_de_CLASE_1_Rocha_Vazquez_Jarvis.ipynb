# Palabras Reservadas
Las palabras reservadas (keywords) corresponden a los nombres de las declaraciones que el interprete de Python incluye por defecto, No se deben utilizar para asignar nombres a otros objetos.

**Ejemplo:**

El siguiente comando desplegara la lista de palabras reservadas:

```python
help("keywords")
```

# El espacio de nombres (namespace)

El espacio de nombres (namespace) contiene una elacion de objetos existentes en la memoria del sistema y los nombres a los que estan ligados.

Python es un lenguaje de alto nivel en el que todos sus elementos son
objetos incluyendo los tipos de datos basicos.
Si un objeto no esta ligado al menos a un nombre, dicho objeto es desechado por el interprete.
Un objeto puede tener más de un nombre.

# El operador de asignación =
Para asignar a un nombre con un objeto, se utiliza el operador de asignación "=" con la siguiente sintaxis.

<nombre> = <objeto>



```python
saludo ="QUE TAL ME LLAVO JARVIS"
print(saludo)
nota="Soy de primer semestre"
print(nota)
carrera="estudio ingenieria en la UAT"
print(carrera)
```
